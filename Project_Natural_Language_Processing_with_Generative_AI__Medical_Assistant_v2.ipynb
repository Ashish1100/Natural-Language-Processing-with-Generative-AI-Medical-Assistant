## Question Answering using LLM with Prompt Engineering and LLM Parameter Tuning

### Query 1: What is the protocol for managing sepsis in a critical care unit?

```python
# Query 1: Sepsis Protocol (Prompt Engineering Section)

system_prompt = """You are a helpful, respectful and honest medical assistant.
Always explain in simple terms for a general audience.
Your answers should not include any harmful, unethical, racist, sexist, toxic, dangerous, or illegal content.
Please ensure that your responses are socially unbiased and positive in nature."""

query_1 = "What is the protocol for managing sepsis in a critical care unit?"

# Prepare model prompt
def prepare_model_prompt(system_prompt, user_query):
    return f"""[INST] <<SYS>>
{system_prompt}
<</SYS>>

{user_query} [/INST]"""

# Generate response
def generate_response(prompt, max_tokens=128, temperature=0, top_p=0.95, top_k=50, repeat_penalty=1.0):
    model_output = llm(
        prompt=prompt,
        max_tokens=max_tokens,
        temperature=temperature,
        top_p=top_p,
        top_k=top_k,
        repeat_penalty=repeat_penalty
    )
    ans = model_output['choices'][0]['text'].strip()
    return ans, model_output

# Execute
model_prompt = prepare_model_prompt(system_prompt, query_1)
ans, moutput = generate_response(model_prompt)
print(ans)
print("completion_tokens =", moutput['usage']['completion_tokens'])
```

## 🔭 My Observations:
### Query 1: Sepsis Protocol (Prompt Engineering)

- **Details:** The code utilizes a detailed system prompt to instruct the LLM to act as a safe, respectful medical assistant and explain the protocol in "simple terms for a general audience."
- **Output Details:** The model successfully outputs a simplified, step-by-step explanation of sepsis management, breaking it down into accessible concepts like "Early recognition" and "Quick response."
- **Important Point:**
  - **Effective Persona Adoption:** The prompt engineering successfully shifted the model's output from highly clinical jargon to an easily digestible format suitable for laypersons.

### Data Chunking

```python
# Data Chunking Function

def get_data_chunks(
    data,
    chunk_size=512,
    chunk_overlap=50,
    model="gpt-3.5-turbo"
):
    """
    Splits documents into chunks using token counts for better LLM compatibility.
    """
    # Initialize tokenizer for precise token counting
    tokenizer = tiktoken.encoding_for_model(model)

    def tiktoken_len(text):
        tokens = tokenizer.encode(text, disallowed_special=())
        return len(tokens)

    text_splitter = RecursiveCharacterTextSplitter(
        chunk_size=chunk_size,
        chunk_overlap=chunk_overlap,
        length_function=tiktoken_len,
        separators=["\n\n", "\n", ". ", " ", ""]
    )

    print(f"Splitting {len(data)} pages into chunks...")
    chunks = text_splitter.split_documents(data)

    # Clean up chunks: ensure metadata exists and filter out very small chunks (less than 20 tokens)
    valid_chunks = []
    for chunk in chunks:
        if tiktoken_len(chunk.page_content) > 20:
            # Ensure metadata reflects the source file
            chunk.metadata["source"] = "medical_diagnosis_manual.pdf"
            valid_chunks.append(chunk)

    print(f"Created {len(valid_chunks)} valid chunks.")
    return valid_chunks

# Execute chunking
texts = get_data_chunks(docs)
```

## 🔭 My Observations:
### Data Chunking

- **Details:** The code defines a custom chunking function using LangChain's `RecursiveCharacterTextSplitter`. Instead of splitting purely by character count, it smartly integrates `tiktoken` to measure chunks by actual token size (setting a target `chunk_size` of 512 tokens with an overlap of 50 tokens).
- **Output Details:** The execution output indicates that the massive 4,114-page document was successfully processed, resulting in a final count of 8,706 valid text chunks ready for embedding.
- **Important Points:**
  - **Token-Aware Chunking:** Using `tiktoken` is a highly effective best practice for RAG. It ensures that the generated chunks will perfectly align with the LLM's actual context window limits, preventing arbitrary cut-offs that happen with standard character splitting.
  - **Built-in Quality Control:** - The logic includes a crucial filtering step that discards any chunk smaller than 20 tokens. This prevents the vector database from being polluted with useless fragments (like orphaned page numbers or single words), which significantly improves retrieval quality.
    - Explicitly appending the `source` metadata ensures traceability, allowing the system to know exactly where a retrieved fact originated from.

### Response Function

```python
def generate_rag_response(user_input,k=3,max_tokens=128,temperature=0,top_p=0.95,top_k=50):
    global qna_system_message,qna_user_message_template
    # Retrieve relevant document chunks
    relevant_document_chunks = retriever.get_relevant_documents(query=user_input,k=k)
    context_list = [d.page_content for d in relevant_document_chunks]

    # Combine document chunks into a single context
    context_for_query = ". ".join(context_list)

    user_message = qna_user_message_template.replace('{context}', context_for_query)
    user_message = user_message.replace('{question}', user_input)

    prompt = qna_system_message + '\n' + user_message

    # Generate the response
    try:
        response = llm(
                  prompt=prompt,
                  max_tokens=max_tokens,
                  temperature=temperature,
                  top_p=top_p,
                  top_k=top_k
                  )

        # Extract and print the model's response
        response = response['choices'][0]['text'].strip()
    except Exception as e:
        response = f'Sorry, I encountered the following error: \n {e}'

    return response
```

```python
def generate_ground_relevance_response(user_input,k=3,max_tokens=128,temperature=0,top_p=0.95,top_k=50):
    global qna_system_message,qna_user_message_template
    # Retrieve relevant document chunks
    relevant_document_chunks = retriever.get_relevant_documents(query=user_input,k=3)
    context_list = [d.page_content for d in relevant_document_chunks]
    context_for_query = ". ".join(context_list)

    # Combine user_prompt and system_message to create the prompt
    prompt = f"""[INST]{qna_system_message}\n
                {'user'}: {qna_user_message_template.format(context=context_for_query, question=user_input)}
                [/INST]"""

    response = llm(
            prompt=prompt,
            max_tokens=max_tokens,
            temperature=temperature,
            top_p=top_p,
            top_k=top_k,
            stop=['INST'],
            )

    answer =  response["choices"][0]["text"]

    # Combine user_prompt and system_message to create the prompt
    groundedness_prompt = f"""[INST]{groundedness_rater_system_message}\n
                {'user'}: {user_message_template.format(context=context_for_query, question=user_input, answer=answer)}
                [/INST]"""

    # Combine user_prompt and system_message to create the prompt
    relevance_prompt = f"""[INST]{relevance_rater_system_message}\n
                {'user'}: {user_message_template.format(context=context_for_query, question=user_input, answer=answer)}
                [/INST]"""

    response_1 = llm(
            prompt=groundedness_prompt,
            max_tokens=max_tokens,
            temperature=temperature,
            top_p=top_p,
            top_k=top_k,
            stop=['INST'],
            )

    response_2 = llm(
            prompt=relevance_prompt,
            max_tokens=max_tokens,
            temperature=temperature,
            top_p=top_p,
            top_k=top_k,
            stop=['INST'],
            )

    return response_1['choices'][0]['text'],response_2['choices'][0]['text']
```

```python
# Query 2: Appendicitis evaluation (Reduced context size)
query2 = """What are the common symptoms for appendicitis, and can it be cured via medicine?
If not, what surgical procedure should be followed to treat it?"""

# function with shorter context
def generate_ground_relevance_response_fixed(user_input, k=2, max_tokens=128):  # Reduced k=2
    global qna_system_message, qna_user_message_template

    # Retrieve FEWER chunks to fit context window
    relevant_document_chunks = retriever.get_relevant_documents(query=user_input, k=2)

    # TRUNCATE context to fit 2300 token limit
    context_list = [d.page_content[:800] for d in relevant_document_chunks]  # Max 800 chars each
    context_for_query = ". ".join(context_list)[:1800]  # Total max 1800 chars

    # Generate RAG answer
    prompt = f"""[INST]{qna_system_message}
    {'user'}: {qna_user_message_template.format(context=context_for_query, question=user_input)}
    [/INST]"""

    response = llm(prompt=prompt, max_tokens=max_tokens, temperature=0, stop=['INST'])
    answer = response["choices"][0]["text"].strip()

    # Groundedness evaluation (SHORTENED prompts)
    groundedness_prompt = f"""[INST]{groundedness_rater_system_message}
    Context: {context_for_query[:1000]}
    Q: {user_input[:200]}
    Answer: {answer[:300]}
    Rate 1-5:[/INST]"""

    relevance_prompt = f"""[INST]{relevance_rater_system_message}
    Q: {user_input[:200]}
    Answer: {answer[:300]}
    Rate 1-5:[/INST]"""

    response_1 = llm(groundedness_prompt, max_tokens=5, stop=['INST'])
    response_2 = llm(relevance_prompt, max_tokens=5, stop=['INST'])

    return response_1['choices'][0]['text'].strip(), response_2['choices'][0]['text'].strip()

groundedness_score, relevance_score = generate_ground_relevance_response_fixed(query2)
print(f"Query: Appendicitis symptoms & treatment")
print(f"Groundedness Score: {groundedness_score}")
print(f"Relevance Score: {relevance_score}")
print("-" * 80)
```

It achieves this by reducing the retriever to `k=2`, aggressively truncating the text of each chunk to 800 characters, and capping the total context size at 1,800 characters.
- **Output Details:** The triple-inference pipeline successfully executes. The LLM evaluator assigns a Groundedness Score of 1 and a Relevance Score of 4. The justification text is cut off almost immediately (e.g., "1: The answer").

# Actionable Insights and Business Recommendations

<font color="teal">**Actionable Technical Insights**</font>

1. **Context Window Management is Critical:** The experiments demonstrated a delicate balance between providing sufficient medical context and exceeding the local model's context window (2,300 tokens). Aggressive truncation (as seen in Query 2) led to a significant drop in factual Groundedness. **Action:** Future iterations should employ advanced retrieval strategies like *Parent Document Retrieval* or *Context Compression* to maximize information density without hitting token limits.
2. **Prompt Engineering Drives Formatting:** The use of Few-Shot and Chain-of-Thought (CoT) prompting significantly improved the structure of the model's output. CoT was particularly effective for complex scenarios (like the leg fracture query), forcing the model to break down its reasoning chronologically.
3. **Automated Evaluation is Viable:** Implementing the "LLM-as-a-judge" framework using strict 1-5 rubrics for Groundedness and Relevance proved highly effective. It successfully flagged instances of data loss (Query 2) and validated high-performing responses (Query 5), providing a scalable way to tune hyperparameters without requiring expensive human-in-the-loop medical auditing for every test run.
4. **Token Parameters Require Tuning:** The consistent truncation of outputs across standard queries highlights that a default `max_tokens=128` is insufficient for comprehensive medical answers. **Action:** Production systems should dynamically adjust `max_tokens` based on the query complexity, or default to a higher baseline (e.g., 512).

3. **Standardization of Care:** By anchoring the AI strictly to vetted, institutional knowledge (like the Merck Manuals) and enforcing high Groundedness scores, hospitals can ensure that all medical staff have instant access to the exact same standardized, up-to-date treatment protocols, reducing variability in patient care.
